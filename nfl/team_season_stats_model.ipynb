```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Model by Season Ranks

### Gathering Data

##### Creating a Ranked Data Frame

```python
df = pd.read_csv('data/all_team_data_trimmed.csv')
df['Season'] = df['Season'].astype(str)
# df.dropna()
df.head()
```

```python
# plotting wins for each column
fig, ax = plt.subplots(9,3,figsize=(18,54))
col_idx = 4
for c in range(9): 
    for r in range(3):
        col = df.columns[col_idx]
        ax[c,r].scatter(df[col], df['Wins'])
        ax[c,r].set_title(col)
        ax[c,r].set_xlabel('Stat')
        ax[c,r].set_ylabel('Wins')
        col_idx += 1

plt.tight_layout()
plt.show()
```

```python
# plotting losses for each column
fig, ax = plt.subplots(9,3,figsize=(18,54))
col_idx = 4
for c in range(9): 
    for r in range(3):
        col = df.columns[col_idx]
        ax[c,r].scatter(df[col], df['Losses'])
        ax[c,r].set_title(col)
        ax[c,r].set_xlabel('Stat')
        ax[c,r].set_ylabel('Losses')
        col_idx += 1

plt.tight_layout()
plt.show()
```
